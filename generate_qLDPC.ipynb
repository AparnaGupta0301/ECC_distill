# Generate Classical LDPC Codes

```python
import os
import subprocess
output_path = '/Users/aparnagupta/Downloads/notebooks/HGP/lib/testing/my_code.pchk'
# Define the parameters
output_file = "_candidate_code.pkl"
ldpc_library_path = "/Users/aparnagupta/Documents/ProtographLDPC/LDPC-library/make-pchk.py"
print_command = f"/Users/aparnagupta/Documents/ProtographLDPC/LDPC-codes/print-pchk {output_path}"
convert_command = f"../../ldpc_codes/LDPC-codes/pchk-to-alist -t {output_path} {output_path}"

# Ensure the output directory exists
#os.makedirs(output_dir, exist_ok=True)

# Command to generate the parity check matrix
generate_command = [
    "python",
    ldpc_library_path,
    "--output-pchk-file", output_file,
    "--code-type", "regular",
    "--construction", "gallager",
    "--n-checks", "6",
    "--n-bits", "8",
    "--checks-per-col", "3",
    # "--seed", "1642"
]

# Generate the LDPC code
print("Generating LDPC code...")
try:
    subprocess.run(generate_command, check=True)
    print("LDPC code generated successfully.")
except subprocess.CalledProcessError as e:
    print(f"Error during LDPC code generation: {e}")

# Print and convert the parity check matrix
print("Printing LDPC code in binary format:")
try:
    subprocess.run(print_command.split(), check=True)
    subprocess.run(convert_command.split(), check=True)
except subprocess.CalledProcessError as e:
    print(f"Error during printing and converting of LDPC code: {e}")
```

```python
import pickle

from scipy.sparse import csr_matrix
import ldpc.code_util

# Load the saved parity check matrix
with open("_candidate_code.pkl", "rb") as file:
    H = pickle.load(file)[0]

# Compute code parameters
n = H.shape[1]
k = ldpc.code_util.compute_code_dimension(H)
d_estimate, number_code_words_sampled, lowest_weight_codewords = ldpc.code_util.estimate_code_distance(H, timeout_seconds=10)

# Display code information
print(f"Code distance estimate, d <= {d_estimate} (no. codewords sampled: {number_code_words_sampled})")
print()
print(f"Code parameters: [n = {n}, k = {k}, d <= {d_estimate}]")
print(f"The code has girth > 4:", not ldpc.code_util.search_cycles(H, girth=4))
```
